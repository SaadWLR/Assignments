```python
crop_clean = crop_df.copy()
crop_clean = crop_clean.dropna(subset=['YIELD_KG'])
crop_clean
```

```python
crop_clean['AREA_ACRES'] = crop_clean['AREA_ACRES'].fillna(crop_df['AREA_ACRES'].mean())
crop_clean
```

```python
crop_clean['DISTRICT'] = crop_clean['DISTRICT'].fillna('Unknown')
crop_clean
```

```python
crop_clean['FERTILIZER_USED'] = crop_clean['FERTILIZER_USED'].fillna('No')
crop_clean
```

```python
crop_clean['YIELD_VALUE'] = crop_clean['YIELD_VALUE'].fillna(crop_clean['YIELD_VALUE'].mean())
crop_clean
```

```python
crop_clean['HARVEST_DATE'] = pd.to_datetime(crop_clean['HARVEST_DATE'], errors='coerce', format='%d/%m/%Y')
crop_clean
```

```python
crop_clean['FARM_ID'] = crop_clean['FARM_ID'].str.replace('F', '').astype(int)
crop_clean
```

```python
crop_clean.isnull().sum()
```

```python
crop_clean.dtypes
```

```python
crop_yield_total = crop_clean.groupby('CROP_TYPE').agg({'YIELD_KG': 'sum'})
crop_yield_total
```

```python
plt.bar(crop_yield_total.index, crop_yield_total['YIELD_KG'], color='green')
plt.title('Total Yield by Crop Type (KG)')
plt.xlabel('Crop Type')
plt.ylabel('Yield (KG)')
plt.show()
```

```python
yield_by_date = crop_clean.pivot_table(values='YIELD_KG', index='HARVEST_DATE', aggfunc='sum')
yield_by_date
```

```python
plt.plot(yield_by_date.index, yield_by_date['YIELD_KG'])
plt.title('Yield Trend Over Time (KG)')
plt.xlabel('Harvest Date')
plt.ylabel('Yield (KG)')
plt.savefig('crop_yield_trend.png')
plt.show()
```
