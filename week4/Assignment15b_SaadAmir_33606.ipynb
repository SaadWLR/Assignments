```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tea = pd.read_csv(r"C:\Users\saada\Downloads\student_tea_snacks _1_.csv")
tea.head()
```

```python
tea.isnull().sum()
```

```python
print(tea['tea_type'].mode()[0])
tea.groupby('tea_type')['student_id'].count()
#green tea
```

```python
tea.groupby('gender')['cups_per_day'].mean()
```

```python
snacks = tea.groupby('snack_choice')['student_id'].count()

plt.bar(snacks.index, snacks.values, color='orange')
plt.xlabel('Snack')
plt.ylabel('Number of Students')
plt.title('Snack Choices')
plt.show()
```

```python
tea.groupby(['age_group', 'prefers_tea'])['student_id'].count()
```

```python
yes_count = tea[tea['prefers_tea'] == 'Yes'].groupby('age_group')['student_id'].count()
total = tea.groupby('age_group')['student_id'].count()

yes_count / total * 100
```
