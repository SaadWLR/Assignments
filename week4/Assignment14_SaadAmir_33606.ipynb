```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\saada\Downloads\pakistani_ecommerce_dataset.csv")
students = pd.read_csv(r"C:\Users\saada\Downloads\student_lifestyle_dataset.csv")
```

```python
print(df['product_category'].mode()[0])
df.groupby('product_category')['user_id'].count()
#highest number of orders means the most popular so home decor
```

```python
df['order_value'].mean()
```

```python
print(df['city'].mode()[0])
df.groupby('city')['user_id'].count() 
```

```python
print(df['payment_method'].mode()[0])
df.groupby('payment_method')['user_id'].count()
```

```python
avg_city = df.groupby('city')['order_value'].mean()
avg_city
```

```python
plt.bar(avg_city.index, avg_city.values, color='green')
plt.xlabel('City')
plt.ylabel('Average Order Value (PKR)')
plt.title('Average Order Value by City')
plt.show()
```
